```python
import pandas as pd
import numpy as np
num_samples = 1000
data = {
    'Material Type': np.random.choice(['Concrete', 'Steel', 'Wood'], num_samples),
    'Cost': np.random.uniform(150, 1500, num_samples),  # Cost between 50 and 500
    'Strength': np.random.uniform(10, 100, num_samples),  # Strength between 10 and 100 MPa
    'Durability': np.random.randint(5, 50, num_samples),  # Durability between 5 and 50 years
    'Environmental Impact': np.random.choice(['Low', 'Medium', 'High'], num_samples),
    'Recyclability': np.random.choice(['Yes', 'No'], num_samples),
}
df = pd.DataFrame(data)
def sustainability_rating(row):
    if row['Environmental Impact'] == 'Low' and row['Recyclability'] == 'Yes':
        return 'High'
    elif row['Environmental Impact'] == 'Medium':
        return 'Medium'
    else:
        return 'Low'

df['Sustainability Rating'] = df.apply(sustainability_rating, axis=1)
df.to_csv('sustainable_materials_dataset.csv', index=False)
print(df.head())
```
